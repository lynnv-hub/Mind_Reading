```python
### Loading the data
import os
import zipfile
import urllib.request
import nibabel as nib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from nilearn import datasets
from nilearn.image import math_img, resample_to_img
from nilearn.maskers import NiftiMasker
from scipy.stats import binomtest, ttest_1samp
from sklearn.model_selection import StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Download and extract the data
DATA_URL = 'https://www.dropbox.com/scl/fi/07hqaz0e8y6moeramv6st/mvpa_assignment_data.zip?rlkey=lcda2vmj1sn49xjok4i71vrvb&dl=1'
DATA_DIR = 'first-level'  # matches the top-level folder inside the zip

class TqdmUpTo(tqdm):
    def update_to(self, blocks=1, block_size=1, total_size=None):
        if total_size is not None:
            self.total = total_size
        self.update(blocks * block_size - self.n)

if not os.path.exists(DATA_DIR):
    with TqdmUpTo(unit='B', unit_scale=True, miniters=1, desc='Downloading') as t:
        urllib.request.urlretrieve(DATA_URL, 'mvpa_data.zip', reporthook=t.update_to)
    with zipfile.ZipFile('mvpa_data.zip', 'r') as z:
        z.extractall('.')  # extract into cwd so the zip's own 'first-level/' becomes DATA_DIR
    print('Data extracted.')
else:
    print('Data directory already exists, skipping download.')

# Load participant list and verify files
manifest_df = pd.read_csv(os.path.join(DATA_DIR, 'selected_subjects.tsv'), sep='\t')
subjects = manifest_df['participant_id'].tolist()
TASK = 'task-faces'

for sub in subjects:
    sub_dir = os.path.join(DATA_DIR, sub)
    for suffix in ['desc-LSA_betas.nii.gz', 'desc-LSA_labels.tsv', 'desc-LSA_mask.nii.gz']:
        fpath = os.path.join(sub_dir, f'{sub}_{TASK}_{suffix}')
        assert os.path.exists(fpath), f'Missing file: {fpath}'

print(f'Loaded {len(subjects)} participants. All files verified.')
```

```python
# Build a pSTS mask from the Harvard-Oxford atlas.
# The STS is a sulcus, so there is no direct label — we approximate it
# by combining the neighbouring gyral labels that form its banks.

ATLAS_NAME = 'cort-maxprob-thr25-2mm'

STS_LABELS = [
    'Superior Temporal Gyrus, posterior division',
    'Middle Temporal Gyrus, posterior division',
    'Middle Temporal Gyrus, temporooccipital part',
]

atlas = datasets.fetch_atlas_harvard_oxford(ATLAS_NAME, symmetric_split=False)
atlas_img = atlas.maps if hasattr(atlas.maps, 'affine') else nib.load(atlas.maps)
labels = list(atlas.labels)

indices = [labels.index(name) for name in STS_LABELS]
atlas_data = atlas_img.get_fdata().astype(np.int32)
mask_data = np.isin(atlas_data, indices).astype(np.uint8)
sts_mask_mni = nib.Nifti1Image(mask_data, atlas_img.affine, atlas_img.header)

print(f'STS mask: {int(mask_data.sum())} voxels')

# To see all available atlas labels (useful for choosing your own ROIs):
for i, name in enumerate(labels):
    print(f'{i:3d}  {name}')
```

```python
### Decoding pipeline: classify positive vs. negative valence within an ROI

POSITIVE = 'positive'
NEGATIVE = 'negative'
CHANCE = 0.5
N_SPLITS = 5
RANDOM_STATE = 42


def decode_subject(sub, roi_mask_mni, classifier=None, n_splits=N_SPLITS, random_state=RANDOM_STATE):
    """Run ROI-based valence decoding for one participant.

    Parameters
    ----------
    sub : str
        Participant ID (e.g., 'sub-0001').
    roi_mask_mni : nibabel.Nifti1Image
        Binary ROI mask in MNI space.
    classifier : sklearn estimator or None
        Classifier to use. If None, defaults to linear SVM (C=1.0).
    n_splits : int
        Number of folds for stratified cross-validation.
    random_state : int
        Random seed for reproducibility.

    Returns
    -------
    dict or None
        Results dictionary, or None if participant is skipped.
    """
    sub_dir = os.path.join(DATA_DIR, sub)
    betas_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_betas.nii.gz'))
    labels_df = pd.read_csv(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_labels.tsv'), sep='\t')
    sub_mask_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_mask.nii.gz'))

    # Resample ROI mask onto the beta grid and intersect with brain mask
    roi_resampled = resample_to_img(roi_mask_mni, betas_img, interpolation='nearest')
    sub_mask_resampled = resample_to_img(sub_mask_img, betas_img, interpolation='nearest')
    combined = math_img('(img1 > 0) & (img2 > 0)', img1=roi_resampled, img2=sub_mask_resampled)

    n_roi_voxels = int(combined.get_fdata().astype(bool).sum())
    if n_roi_voxels == 0:
        warnings.warn(f'Skipping {sub}: empty ROI after mask intersection')
        return None

    # Extract feature matrix: (n_trials, n_voxels)
    masker = NiftiMasker(mask_img=combined, standardize=False)
    X = masker.fit_transform(betas_img)

    # Keep only positive and negative trials
    keep = labels_df['valence'].isin([POSITIVE, NEGATIVE]).values
    X = X[keep]
    y = labels_df.loc[keep, 'valence'].values
    n_pos = int((y == POSITIVE).sum())
    n_neg = int((y == NEGATIVE).sum())

    if min(n_pos, n_neg) < n_splits:
        warnings.warn(f'Skipping {sub}: too few trials per class for {n_splits}-fold CV')
        return None

    # Classification
    if classifier is None:
        classifier = SVC(kernel='linear', C=1.0)

    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', classifier),
    ])
    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    fold_scores = cross_val_score(pipe, X, y, cv=cv, scoring='accuracy')
    predictions = cross_val_predict(pipe, X, y, cv=cv)

    accuracy = float(fold_scores.mean())
    n_correct = int((predictions == y).sum())
    binom_p = binomtest(n_correct, len(y), p=CHANCE, alternative='greater').pvalue

    return {
        'subject': sub,
        'n_trials': len(y),
        'n_positive': n_pos,
        'n_negative': n_neg,
        'n_roi_voxels': n_roi_voxels,
        'accuracy': accuracy,
        'accuracy_std': float(fold_scores.std(ddof=1)),
        'binomial_p': float(binom_p),
    }


# Run decoding in the pSTS across all participants
results = []
for i, sub in enumerate(subjects, 1):
    result = decode_subject(sub, sts_mask_mni)
    if result is None:
        print(f'[{i}/{len(subjects)}] {sub}: SKIPPED')
        continue
    results.append(result)
    print(f'[{i}/{len(subjects)}] {sub}: acc={result["accuracy"]:.3f}  p={result["binomial_p"]:.3f}')

results_df = pd.DataFrame(results)

# Group-level test
mean_acc = results_df['accuracy'].mean()
sd_acc = results_df['accuracy'].std(ddof=1)
n_above = int((results_df['binomial_p'] < 0.05).sum())
t_res = ttest_1samp(results_df['accuracy'], CHANCE, alternative='greater')

print(f'\nGroup mean accuracy: {mean_acc:.3f} (SD {sd_acc:.3f})')
print(f'Above chance (binomial p<0.05): {n_above}/{len(results_df)}')
print(f'Group t-test vs {CHANCE}: t={t_res.statistic:.3f}, df={len(results_df)-1}, p={t_res.pvalue:.4f}')
```

```python
# Example visualisation: per-participant accuracy in the pSTS
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(results_df['accuracy'].values, 'o', color='steelblue', alpha=0.6, label='Individual participants')

sem = sd_acc / np.sqrt(len(results_df))
ax.axhline(mean_acc, color='steelblue', linewidth=2, label=f'Group mean = {mean_acc:.3f}')
ax.axhspan(mean_acc - sem, mean_acc + sem, color='steelblue', alpha=0.15)
ax.axhline(CHANCE, color='grey', linestyle='--', linewidth=1, label='Chance (0.5)')

ax.set_xlabel('Participant')
ax.set_ylabel('Decoding accuracy (5-fold CV)')
ax.set_title('Emotional Valence Decoding in the pSTS')
ax.legend(loc='upper right')
ax.set_ylim(0.0, 1.0)
plt.tight_layout()
plt.show()
```

```python
def build_mask(atlas_name, region_labels):
    atlas = datasets.fetch_atlas_harvard_oxford(atlas_name, symmetric_split=False)
    atlas_img = atlas.maps if hasattr(atlas.maps, 'affine') else nib.load(atlas.maps)
    labels = list(atlas.labels)

    indices = [labels.index(name) for name in region_labels]
    atlas_data = atlas_img.get_fdata().astype(np.int32)

    mask_data = np.isin(atlas_data, indices).astype(np.uint8)
    mask_img = nib.Nifti1Image(mask_data, atlas_img.affine, atlas_img.header)

    return mask_img, int(mask_data.sum())

FFA_LABELS = [
    'Temporal Fusiform Cortex, posterior division',
    'Temporal Occipital Fusiform Cortex',
]

ACC_LABELS = [
    'Cingulate Gyrus, anterior division',
    'Paracingulate Gyrus',
]

AMYGDALA_LABELS = [
    'Left Amygdala',
    'Right Amygdala',
]

ffa_mask_mni, ffa_voxels = build_mask('cort-maxprob-thr25-2mm', FFA_LABELS)
acc_mask_mni, acc_voxels = build_mask('cort-maxprob-thr25-2mm', ACC_LABELS)
amygdala_mask_mni, amygdala_voxels = build_mask('sub-maxprob-thr25-2mm', AMYGDALA_LABELS)

print(f'FFA mask: {ffa_voxels} voxels')
print(f'ACC mask: {acc_voxels} voxels')
print(f'Amygdala mask: {amygdala_voxels} voxels')
```

```python
def balance_trials(labels_df, random_state=42):
    np.random.seed(random_state)

    pos_emotions = ['joy', 'pride']
    neg_emotions = ['anger', 'contempt']

    selected_idx = []

    for pos, neg in zip(pos_emotions, neg_emotions):
        pos_idx = labels_df[labels_df['trial_type'] == pos].index.values
        neg_idx = labels_df[labels_df['trial_type'] == neg].index.values

        n = min(len(pos_idx), len(neg_idx))

        pos_sample = np.random.choice(pos_idx, n, replace=False)
        neg_sample = np.random.choice(neg_idx, n, replace=False)

        selected_idx.extend(pos_sample)
        selected_idx.extend(neg_sample)

    return np.array(sorted(selected_idx))
```

```python
from sklearn.model_selection import LeaveOneOut

def decode_subject_loo(sub, roi_mask_mni, classifier):

    sub_dir = os.path.join(DATA_DIR, sub)

    betas_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_betas.nii.gz'))
    labels_df = pd.read_csv(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_labels.tsv'), sep='\t')
    sub_mask_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_mask.nii.gz'))

    roi_resampled = resample_to_img(roi_mask_mni, betas_img, interpolation='nearest')
    sub_mask_resampled = resample_to_img(sub_mask_img, betas_img, interpolation='nearest')

    combined = math_img('(img1 > 0) & (img2 > 0)',
                        img1=roi_resampled,
                        img2=sub_mask_resampled)

    if combined.get_fdata().sum() == 0:
        return None

    masker = NiftiMasker(mask_img=combined, standardize=True)
    X = masker.fit_transform(betas_img)

    keep_idx = balance_trials(labels_df)
    X = X[keep_idx]
    y = labels_df.loc[keep_idx, 'valence'].values

    y = np.array([1 if v == 'positive' else 0 for v in y])

    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('clf', classifier),
    ])

    loo = LeaveOneOut()

    correct = 0
    total = 0

    for train_idx, test_idx in loo.split(X):

        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]

        pipe.fit(X_train, y_train)
        pred = pipe.predict(X_test)

        correct += int(pred[0] == y_test[0])
        total += 1

    if total == 0:
        return None

    return correct / total
```

```python
#second classifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
svm_clf = SVC(kernel='linear', C=1.0)
logreg_clf = LogisticRegression(max_iter=1000)
```

```python
rois = {
    'FFA': ffa_mask_mni,
    'ACC': acc_mask_mni,
    'Amygdala': amygdala_mask_mni
}

classifiers = {
    'SVM': svm_clf,
    'LogReg': logreg_clf
}

results = []

for sub in subjects:
    for roi_name, roi_mask in rois.items():
        for clf_name, clf in classifiers.items():

            acc = decode_subject_loo(sub, roi_mask, clf)

            results.append({
                'subject': sub,
                'roi': roi_name,
                'classifier': clf_name,
                'accuracy': acc
            })

results_df = pd.DataFrame(results)
```

```python
from nilearn import plotting
from nilearn.image import mean_img

def compute_svm_weight_map(sub, roi_mask_mni):

    sub_dir = os.path.join(DATA_DIR, sub)

    betas_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_betas.nii.gz'))
    labels_df = pd.read_csv(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_labels.tsv'), sep='\t')
    sub_mask_img = nib.load(os.path.join(sub_dir, f'{sub}_{TASK}_desc-LSA_mask.nii.gz'))

    # Resample masks
    roi_resampled = resample_to_img(roi_mask_mni, betas_img, interpolation='nearest')
    sub_mask_resampled = resample_to_img(sub_mask_img, betas_img, interpolation='nearest')

    combined = math_img(
        '(img1 > 0) & (img2 > 0)',
        img1=roi_resampled,
        img2=sub_mask_resampled
    )

    if combined.get_fdata().sum() == 0:
        return None

    masker = NiftiMasker(mask_img=combined, standardize=True)

    X = masker.fit_transform(betas_img)

    keep_idx = balance_trials(labels_df)

    X = X[keep_idx]
    y = labels_df.loc[keep_idx, 'valence'].values

    y = np.array([1 if v == 'positive' else 0 for v in y])

    scaler = StandardScaler()
    X = scaler.fit_transform(X)

    clf = SVC(kernel='linear', C=1.0)

    clf.fit(X, y)

    weights = clf.coef_[0]

    weight_img = masker.inverse_transform(weights)

    return weight_img

roi_weight_maps = {}

for roi_name, roi_mask in rois.items():

    subject_maps = []

    for sub in subjects:

        try:
            wmap = compute_svm_weight_map(sub, roi_mask)

            if wmap is not None:
                subject_maps.append(wmap)

        except Exception as e:
            print(f'{sub} failed for {roi_name}: {e}')

    avg_map = mean_img(subject_maps)

    roi_weight_maps[roi_name] = avg_map

general_caption = (
    "Average SVM weight map across 30 participants. "
    "Red voxels indicate positive valence weights, "
    "blue voxels indicate negative valence weights."
)

roi_captions = {
    "FFA": (
        "FFA: Weights cluster in bilateral fusiform and occipitotemporal cortex. "
        "Right hemisphere shows stronger positive weights, consistent with face processing lateralization."
    ),
    "ACC": (
        "ACC: Weights are centered in anterior cingulate and paracingulate cortex. "
        "A dorsal–ventral gradient suggests subregional differences in valence coding."
    ),
    "Amygdala": (
        "Amygdala: Strong bilateral clusters with highest weight magnitude. "
        "Some hemispheric asymmetry appears in valence-related weighting."
    )
}

for roi_name, weight_map in roi_weight_maps.items():

    plotting.plot_stat_map(
        weight_map,
        title=f'{roi_name} average SVM weight map',
        display_mode='z',
        cut_coords=6,
        colorbar=True
    )

    plt.figtext(
        0.5,
        -0.05,
        roi_captions[roi_name],
        wrap=True,
        ha='center',
    )

    # General caption (shared info)
    plt.figtext(
        0.5,
        -0.12,
        general_caption,
        wrap=True,
        ha='center',
    )

    plotting.show()
```

```python
#Summary statistics table
def cohens_d(accs, chance=0.5):
    return (accs.mean() - chance) / accs.std(ddof=1)

summary = []

for roi in roi_order:
    for clf in clf_order:
        subset = results_df[
            (results_df['roi'] == roi) &
            (results_df['classifier'] == clf)
        ]

        accs = subset['accuracy'].values

        t, p = ttest_1samp(accs, 0.5, alternative='greater')

        summary.append({
            'ROI': roi,
            'Classifier': clf,
            'Mean': accs.mean(),
            'SD': accs.std(ddof=1),
            't': t,
            'p': p,
            'Cohens_d': cohens_d(accs)
        })

summary_df = pd.DataFrame(summary)

# Bonferroni correction (6 tests)
summary_df['p_corrected'] = (summary_df['p'] * 6).clip(upper=1.0)

summary_df
```

```python
#Count significant participants (binomial)
from scipy.stats import binomtest

binomial_results = []

for roi in roi_order:
    for clf in clf_order:

        subset = results_df[
            (results_df['roi'] == roi) &
            (results_df['classifier'] == clf)
        ]

        # count above chance per participant
        successes = np.sum(subset['accuracy'] > 0.5)
        n = len(subset)

        p_val = binomtest(successes, n, p=0.5, alternative='greater').pvalue

        binomial_results.append({
            'ROI': roi,
            'Classifier': clf,
            'Above_chance_subjects': successes,
            'Total': n,
            'p': p_val
        })

binomial_df = pd.DataFrame(binomial_results)
binomial_df
```

**Interpretation**

The decoding analysis showed that emotional valence could only be classified slightly above chance level across all three regions of interest. Mean decoding accuracies ranged from approximately 50.5% to 51.9%, depending on the ROI and classifier. Because chance level for binary classification was 50%, these results indicate that the voxel activity patterns within these regions contained very limited information about emotional valence. Although some ROIs showed numerically higher accuracies than others, none of the decoding results survived statistical testing after Bonferroni correction.

The strongest decoding performance was observed in the amygdala using the linear SVM classifier, with a mean accuracy of 51.9% (SD = 0.145). However, this effect was not statistically significant, t(29) = 0.74, p = .234, corrected p = 1.0. The FFA showed very similar performance, with mean accuracies of 51.7% for the SVM and 51.5% for logistic regression. This suggests that even though some decoding performance was numerically above chance, the magnitude of the effects was weak and highly variable across participants.

The binomial tests further supported this conclusion. Across ROIs and classifiers, only 13-16 out of 30 participants showed above chance decoding. **Reflection**

One of the main challenges during implementation was debugging the Leave One Out cross validation procedure. Early versions of the implementation produced empty or invalid accuracy values because folds were accidentally skipped during evaluation. The accurarcy was coming up at `NaN` because it never got computed, which means scores were staying empty or becoming invalid. Then I tried to return `np.mean(scores)` which was always `NaN`. I fixed this by going over how Leave One Out works in practice and ensuring that predictions were generated correctly for every fold, as my original version was skipping folds and not storing means correctly.

The final analysis differed from the original proposal in a few ways. The proposal initially described balancing emotional categories within each cross validation fold and holding out one positive and one negative trial together during Leave One Out validation. In my final version, balancing occurred before cross validation through random subsampling of emotional categories rather than separately within each fold. In addition, the implemented Leave One Out procedure held out a single trial per fold instead of a positive negative pair. Although this differed from the original plan, it still maximized training data usage and remained appropriate given the small number of trials available. Another change was the inclusion of logistic regression as an additional classifier following feedback that classifier performance should be compared directly.

The approach also has several important limitations. First, the number of trials per participant was relatively small, which limits the stability of decoding estimates and increases sensitivity to noise. fMRI data are inherently noisy, and small trial counts make it more difficult to distinguish meaningful signal from random variation. Second, the analysis relied on predefined ROIs based on anatomical atlases. Although this simplifies interpretation, anatomical ROIs may not align perfectly with the functional organization of emotional processing regions in individual participants. Third, the use of linear classifiers assumes that the relationship between voxel patterns and emotional valence is approximately linear. While linear models are appropriate for small datasets and reduce overfitting risk, they may fail to capture more complex representational structures. Fourth, averaging classifier weight maps across participants may obscure meaningful individual differences in neural representation.

This project demonstrated several important aspects of multivariate neuroimaging analysis. One major insight was that MVPA can detect distributed information patterns even when traditional univariate analyses might fail to find large activation differences. Rather than focusing on whether a single voxel is more active for one condition, MVPA evaluates whether patterns across many voxels collectively contain discriminative information. The project also showed how methodological choices such as ROI definition, cross validation strategy, and classifier selection can substantially affect analysis outcomes and interpretation. Additionally, the project highlighted the importance of careful preprocessing, balancing procedures, and debugging in machine learning analyses of neuroimaging data. Small implementation errors can strongly influence results, especially with limited data. Overall, the assignment provided practical experience with the strengths and limitations of MVPA and demonstrated how distributed brain activity patterns can be used to study emotional processing in social perception tasks.
